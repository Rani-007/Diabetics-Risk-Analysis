<h2 style="color:#1976D2; text-decoration:underline;">
1. Creating Age Group
</h2>

```python
def age_group(age):
    if age <= 30:
        return '18-30'
    elif age <= 45:
        return '31-45'
    elif age <= 60:
        return '46-60'
    elif age <= 75:
        return '61-75'
    else:
        return '76+'

df['Age_Group'] = df['Age'].apply(age_group)
```

```python
df[['Age','Age_Group']].head(10)
```

```python
df['Age_Group'].value_counts().sort_index()
```

## Interpretation
#### The patients were distributed across five age groups ,with the highest number belonging to the 46-60 age group

<h2 style="color:#1976D2; text-decoration:underline;">
2. Creating BMI Category
</h2>

```python
def bmi_category(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif bmi < 25:
        return 'Normal'
    elif bmi < 30:
        return 'Overweight'
    else:
        return 'Obese'

df['BMI_Category'] = df['BMI'].apply(bmi_category)
```

```python
df[['BMI','BMI_Category']].head(10)
```

```python
df['BMI_Category'].value_counts()
```

<h4 style="color:#4CAF50;">Interpretation</h4>

The BMI values were categorized into four groups—Underweight, Normal, Overweight, and Obese—using conditional statements.
This derived feature makes it easier to compare diabetes risk across different BMI categories.from this we can say the obese category has the highest
number of people.

<h2 style="color:#F44336;">Outlier Detection Using IQR</h2>

```python
numeric_columns=df.select_dtypes(include=['int64','float64']).columns

print(numeric_columns)
```

```python
for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    
    print(column, ":", len(outliers), "outliers")
```

```python
Q1 = df['Diabetes_Risk_Score'].quantile(0.25)
Q3 = df['Diabetes_Risk_Score'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
print(df['Diabetes_Risk_Score'].min())
print(df['Diabetes_Risk_Score'].max())
```

```python
outliers = df[
    (df['Diabetes_Risk_Score'] < lower_bound) |
    (df['Diabetes_Risk_Score'] > upper_bound)
]

outliers['Diabetes_Risk_Score'].sort_values()
```

<h4 style="color:#4CAF50;">Interpretation</h4>

#### The IQR method was used to identify potential outliers in the numerical features. For Diabetes_Risk_Score, Q1 was 64 and Q3 was 86, giving an 
#### IQR of 22. The lower and upper bounds were 31 and 119, respectively. A total of 237 observations were identified as potential outliers, with scores
#### ranging from 13 to 30. Since these values fall within the valid observed range of the risk score (13–100) and may represent genuine low-risk individuals,
####  they were retained in the dataset rather than removed.

<h2 style="color:#E91E63; text-decoration:underline;">11. Age Group and BMI Category vs Diabetes Risk</h2>
